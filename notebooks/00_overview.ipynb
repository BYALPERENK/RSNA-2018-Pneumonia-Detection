# 00 · Project Overview

**Competition:** [RSNA Pneumonia Detection Challenge](https://www.kaggle.com/competitions/rsna-pneumonia-detection-challenge/overview) (Kaggle, 2018)

This project rebuilds a solution to the challenge from scratch as a learning (and maybe portfolio) project. The notebooks in this folder are meant to be read in order and cover the full pipeline, from data exploration to the final submission.

## The competition

- **Host:** Radiological Society of North America (RSNA), with the US National Institutes of Health (NIH), the Society of Thoracic Radiology and MD.ai.
- **Dates:** August 27 – October 31, 2018, in two stages. Stage 2 released a new test set in its final week.
- **Scale:** 1,499 teams, $30,000 in prizes.

Pneumonia is a leading cause of death, especially among young children. On a chest X-ray it usually shows up as one or more areas of increased **lung opacity**. Reading these images is hard: other conditions (fluid, bleeding, collapsed lung, cancer) look similar, and radiologists must read a large number of images every shift. The goal of the challenge was an algorithm that screens chest X-rays and flags likely pneumonia cases so they can be reviewed first.

## The task

The input is a single frontal chest X-ray (DICOM, 1024 × 1024, grayscale). The model has to **locate lung opacities** that suggest pneumonia and return a bounding box for each one:

- An image can have **zero, one or several** boxes.
- Images without pneumonia must get **no boxes at all**. This includes many images that are abnormal for other reasons.
- Each box is given as `confidence x y width height` in pixels.

So this is an **object detection** problem, but it also hides an image-level **classification** problem: deciding whether an image should have any boxes at all matters as much as placing the boxes correctly.

## Evaluation metric

Submissions are scored with **mean average precision over IoU thresholds** from 0.40 to 0.75 in steps of 0.05. For each image and each threshold $t$:

$$\text{precision}(t) = \frac{TP(t)}{TP(t) + FP(t) + FN(t)}$$

The image score is the mean over the 8 thresholds, and the final score is the mean over all test images.

One rule shapes the whole competition: **if an image has no ground-truth boxes, any predicted box gives that image a score of 0.** An image with no boxes and no predictions is skipped, so it does not change the score. False positives on healthy images are therefore very costly.

## Data

| Split | Images | Labels |
|-------|-------:|--------|
| Train | 26,684 | Boxes (`stage_2_train_labels.csv`) and a 3-class label (`stage_2_detailed_class_info.csv`) |
| Test  |  3,000 | None (predictions are scored on Kaggle) |

The 3-class label is `Lung Opacity`, `No Lung Opacity / Not Normal` or `Normal`. Only `Lung Opacity` images have boxes. The data is not included in this repository; see the README for how to download it.
